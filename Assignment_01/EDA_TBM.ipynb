```python
# use median imputation for all columns with few missing values and numerical data:
X['Age'] = X['Age'].fillna(X['Age'].median())
X['BMI'] = X['BMI'].fillna(X['BMI'].median())
X['Height'] = X['Height'].fillna(X['Height'].median())
X['Weight'] = X['Weight'].fillna(X['Weight'].median())
X['Length_of_Stay'] = X['Length_of_Stay'].fillna(X['Length_of_Stay'].median())
X['Alvarado_Score'] = X['Alvarado_Score'].fillna(X['Alvarado_Score'].median())
X['Paedriatic_Appendicitis_Score'] = X['Paedriatic_Appendicitis_Score'].fillna(X['Paedriatic_Appendicitis_Score'].median())
X['Body_Temperature'] = X['Body_Temperature'].fillna(X['Body_Temperature'].median())
X['WBC_Count'] = X['WBC_Count'].fillna(X['WBC_Count'].median())
X['Neutrophil_Percentage'] = X['Neutrophil_Percentage'].fillna(X['Neutrophil_Percentage'].median())
X['RBC_Count'] = X['RBC_Count'].fillna(X['RBC_Count'].median())
X['Hemoglobin'] = X['Hemoglobin'].fillna(X['Hemoglobin'].median())
X['RDW'] = X['RDW'].fillna(X['RDW'].median())
X['Thrombocyte_Count'] = X['Thrombocyte_Count'].fillna(X['Thrombocyte_Count'].median())
X['CRP'] = X['CRP'].fillna(X['CRP'].median())
```
